```python
# ============================================
# STAGE 2: Crop ROI → DINOv2 OA Grading
# ============================================

import cv2
import torch
import numpy as np
from PIL import Image
from transformers import AutoImageProcessor, AutoModel
import torch.nn as nn
from torchvision import transforms

# Load DINOv2
processor = AutoImageProcessor.from_pretrained('facebook/dinov2-base')
backbone = AutoModel.from_pretrained('facebook/dinov2-base')

# Add classification head (4 OA grades: 0,1,2,3)
class OAClassifier(nn.Module):
    def __init__(self, backbone, num_classes=4):
        super().__init__()
        self.backbone = backbone
        self.classifier = nn.Sequential(
            nn.Linear(768, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

    def forward(self, pixel_values):
        outputs = self.backbone(pixel_values=pixel_values)
        cls_token = outputs.last_hidden_state[:, 0, :]  # CLS token
        return self.classifier(cls_token)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
oa_model = OAClassifier(backbone, num_classes=4).to(device)
print("OA Classifier ready ✅")
```

```python
# ============================================
# FULL PIPELINE: YOLO → Crop → OA Grade
# ============================================

def predict_oa(image_path, yolo_model, oa_model, processor, device):

    # --- Step 1: YOLO detects bone region ---
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    results = yolo_model(image_path, conf=0.25)
    boxes = results[0].boxes.xyxy.cpu().numpy()

    if len(boxes) == 0:
        print("No bone detected")
        return None

    # Take highest confidence box
    box = boxes[0]
    x1, y1, x2, y2 = map(int, box)

    # --- Step 2: Crop ROI ---
    cropped = img_rgb[y1:y2, x1:x2]
    cropped_pil = Image.fromarray(cropped)

    # --- Step 3: DINOv2 OA Classification ---
    inputs = processor(images=cropped_pil, return_tensors="pt").to(device)

    oa_model.eval()
    with torch.no_grad():
        logits = oa_model(inputs['pixel_values'])
        pred = torch.argmax(logits, dim=1).item()

    grade_map = {0: 'Grade 0 - Normal', 1: 'Grade 1 - Mild',
                 2: 'Grade 2 - Moderate', 3: 'Grade 3 - Severe'}

    print(f"Detected OA: {grade_map[pred]}")

    # Visualize
    cv2.rectangle(img, (x1,y1), (x2,y2), (0,255,0), 2)
    cv2.putText(img, grade_map[pred], (x1, y1-10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)

    result_img = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    display(result_img)

    return pred

# Run pipeline
yolo_model = YOLO(r"E:\ML\Labeled data\yolov8\runs\detect\ddh_detection\weights\best.pt")
predict_oa(sample, yolo_model, oa_model, processor, device)
```
